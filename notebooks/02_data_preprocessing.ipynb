# Seperate data into played games (score already determined) and 2026 WC fixtures

```python
historical_matches = results_1872_2026[results_1872_2026['home_score'].notna()].copy()
wc_fixtures = results_1872_2026[results_1872_2026['home_score'].isna()].copy()

historical_matches.head()
```

```python
# save the data into a seperate csv file
historical_matches.to_csv("../data/processed/historical_matches.csv")
wc_fixtures.to_csv("../data/processed/wc_fixtures.csv")

print(f"Historical Matches: {len(historical_matches)}")
print(f"2026 WC Fixtures: {len(wc_fixtures)}")
```
